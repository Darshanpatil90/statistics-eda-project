# Step 8: Statistical Analysis

```python
# Correlation Matrix
df.corr(numeric_only=True)
```

### Observation

The correlation matrix shows the relationship between numerical variables.

```python
Q1 = df["balance"].quantile(0.25)
Q3 = df["balance"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df["balance"] < lower) | (df["balance"] > upper)]

print("Number of Outliers:", len(outliers))
```

### Observation

The IQR method is used to detect outliers in the balance column.

```python
from scipy.stats import ttest_ind

male = df[df["gender"]=="Male"]["balance"]
female = df[df["gender"]=="Female"]["balance"]

t_stat, p_value = ttest_ind(male, female)

print("T-Statistic:", t_stat)
print("P-Value:", p_value)
```

### Observation

The T-test compares the average balance of male and female customers.

```python
from scipy.stats import f_oneway

france = df[df["country"]=="France"]["balance"]
germany = df[df["country"]=="Germany"]["balance"]
spain = df[df["country"]=="Spain"]["balance"]

f_stat, p_value = f_oneway(france, germany, spain)

print("F-Statistic:", f_stat)
print("P-Value:", p_value)
```

### Observation

ANOVA compares the average balance among customers from France, Germany, and Spain.

```python
from scipy.stats import chi2_contingency

table = pd.crosstab(df["gender"], df["churn"])

chi2, p, dof, expected = chi2_contingency(table)

print("Chi-Square:", chi2)
print("P-Value:", p)
```

### Observation

The Chi-Square test checks whether gender and churn are related.
